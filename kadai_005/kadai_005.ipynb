```python
import numpy as np
import matplotlib.pyplot as plt

nrows = 2
ncols = 3
fig, axes = plt.subplots(nrows, ncols, tight_layout=True)

# データを作成
# 折れ線グラフ
y0 = [1, 2, -5, 2]
axes[0,0].plot(y0)

# sin関数
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
axes[0,1].plot(x, y1)

# ヒストグラム
y2 = np.random.randn(1000)
axes[0,2].hist(y2)

# 散布図
x = np.random.randn(100)
y = np.random.randn(100)
axes[1,0].scatter(x, y)

plt.show()
```
