```python
category_df = pd.read_csv('category.csv')
category_df
```

```python
df3 = pd.merge(df, category_df[['商品番号', 'カテゴリー']], how='inner', on='商品番号')
df3
```

```python
category_counts = df3['カテゴリー'].value_counts()
category_counts
```

```python
def plot_category_frequency(counts: pd.Series):
    fig, ax = plt.subplots(figsize=(8, 5))
    counts.plot(kind='bar')

    ax.set_xlabel('カテゴリー', fontsize=12)
    ax.set_ylabel('出現頻度', fontsize=12)
    plt.show()
```

```python
plot_category_frequency(category_counts)
```
